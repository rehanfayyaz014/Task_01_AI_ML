```python
## 1. Importing Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load data set
df = pd.read_csv("punjab-crime-stats.csv")
#1st five rows
print(df.head())
```

```python
#identifyin colimns types
print("Numerical Columns:")
print(df.select_dtypes(include=np.number).columns.tolist())

print("Categorical Columns:")
print(df.select_dtypes(include="object").columns.tolist())
```

```python
#Data Cleaning
missing_vlaues=df.isnull().sum()
print(missing_vlaues)
```

```python
#Dupicate data
print("Dulicate rows : ", df.duplicated().sum())
```

```python
#Checking inviald values
print("Negative Crime count : ", (df["CrimeCount"]<0).sum())
print("Negative Population count : ", (df["Population"]<0).sum())
print("Invaild Years : ", (df["Year"]<0).sum())
```

```python
#STATISTICAL ANALYSIS
```

```python
correlation = df[["Year", "CrimeCount", "Population"]].corr()

correlation
```

```python
# Visualization 1 — Crime Count Distribution
plt.figure(figsize=(10, 6))
plt.hist(df["CrimeCount"], bins=30)

plt.title("Distrbution of Crime Counts")
plt.xlabel("Crime Count")
plt.ylabel("Frequency")

plt.show()
```

```python
'''This grapgh show the totoal reported crimes count among top 10 dist'''
district_crime = (
    df.groupby("District")["CrimeCount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

district_crime.plot(kind="bar")

plt.title("Top 10 Distrcts by Total Crime Count")
plt.xlabel("District")
plt.ylabel("Total Crime Count")

plt.show()
```

```python
#The pie chart shows the percentage contribution of the major crime types in the dataset.
crime_counts = df["CrimeType"].value_counts().head(8)

plt.figure(figsize=(8, 8))

plt.pie(
    crime_counts,
    labels=crime_counts.index,
    autopct="%1.1f%%"
)

plt.title("Distribution of Major Crime Types")

plt.show()
```

```python
""""The scatter plot shows the relationship between population
and reported crime count. It helps us examine whether areas
with larger populations tend to have higher reported crime counts."""
plt.figure(figsize=(10, 6))

plt.scatter(
    df["Population"],
    df["CrimeCount"],
    alpha=0.5
)

plt.title("Population vs Crime Count")
plt.xlabel("Population")
plt.ylabel("Crime Count")

plt.show()
```

```python
""""The line chrt show how the total reported crime count changed
over the years. It help identify increasing, decreasing, or
flutuating crime trends over time."""

yearly_crime = df.groupby("Year")["CrimeCount"].sum()

plt.figure(figsize=(10, 6))

plt.plot(
    yearly_crime.index,
    yearly_crime.values,
    marker="o"
)

plt.title("Total Crime Count Over Time")
plt.xlabel("Year")
plt.ylabel("Total Crime Count")

plt.show()
```

```python
"""""The box plot shows the spread of crime counts and helps identify
potential outliers. Values far outside the main distribution may
represent unusually high crime counts."""
plt.figure(figsize=(10, 6))

sns.boxplot(x=df["CrimeCount"])

plt.title("Box Plot of Crime Count")
plt.xlabel("Crime Count")

plt.show()
```

```python

"""The correlation heatmap shows the strength and direction of
relationships between numerical variables. Values closer to +1
represent stronger positive relationships, while values closer
to -1 represent stronger negative relationships."""

plt.figure(figsize=(8, 6))

correlation = df[["Year", "CrimeCount", "Population"]].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Corelation Heatmap")

plt.show()
```

```python
""""The count plot compares the frequency
 of differnt crime type.
"""
plt.figure(figsize=(12, 7))

sns.countplot(
    data=df,
    y="CrimeType",
    order=df["CrimeType"].value_counts().index
)

plt.title("Frequncy of Crime Typed")
plt.xlabel("Numbr of Record")
plt.ylabel("Crime Type")

plt.show()
```

```python
"""The pair plot provides a combned view of relationship between
numercal variable."""

numeric_data = df[["Year", "CrimeCount", "Population"]].dropna()

sns.pairplot(numeric_data)

plt.show()
```

The violin plot compres the distribution and spread of crime
count across different crime type.

```python
plt.figure(figsize=(12, 7))

sns.violinplot(
    data=df,
    x="CrimeType",
    y="CrimeCount"
)

plt.title("Crime Count Distribution by Crime Type")
plt.xlabel("Crime Type")
plt.ylabel("Crime Count")
plt.xticks(rotation=45)

plt.show()
```
